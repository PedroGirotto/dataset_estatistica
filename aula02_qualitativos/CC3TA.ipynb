```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df1 = pd.read_csv("index_1.csv")
df1.drop("card", axis=1, inplace=True)
df2 = pd.read_csv("index_2.csv")
```

```python
df1.head(5)
```

```python
df1.tail(1)
```

```python
display(df1["coffee_name"].value_counts())

plt.subplots(figsize=(14,12))
sns.countplot(data=df1, x="coffee_name")
plt.xticks(rotation=60)
plt.show()
```

```python
valores = df1["coffee_name"].value_counts().values
categoria = df1["coffee_name"].value_counts().index
plt.subplots(figsize=(14,12))
plt.pie(valores, labels=categoria, labeldistance=1.15)
plt.show()
```

```python
total = df1.shape[0]
df1["coffee_name"].value_counts()/total
# df1["coffee_name"].value_counts(normalize=True)
```

```python
df1["date"] = pd.to_datetime(df1["date"])
```

```python
df1["day"] = df1["date"].apply(lambda x: x.day_name())
```

```python
df1.groupby(["day", "coffee_name"])["cash_type"].count().reset_index()
```

```python
fig, axs = plt.subplots(figsize=(20,14), ncols=2)
sns.countplot(data=df1, x="day", ax=axs[0])
sns.countplot(data=df1, x="day", hue = "coffee_name", ax=axs[1])
plt.show()
```

```python
df1["year"] = df1["date"].apply(lambda x: x.year)
```

```python
df1["month"] = df1["date"].apply(lambda x: x.month_name())
```

```python
data = df1[(df1["month"]!="March") | (df1["year"]!=2025)]

fig, axs = plt.subplots(figsize=(20,14), ncols=2)
plt.subplot(1,2,1)
sns.countplot(data=data, x="month", ax=axs[0])
plt.xticks(rotation=60)

plt.subplot(1,2,2)
sns.countplot(data=data, x="month", hue = "coffee_name", ax=axs[1])
plt.xticks(rotation=60)
plt.show()
```
